```python
import pandas as pd

historical_df = pd.read_csv("../data/nq_first_to_100_predictions_evaluated.csv")
live_df = pd.read_csv("../data/weekly_trade_log_2026_week_5_with_auction_direction.csv")

print("Historical:", historical_df.shape)
print("Live:", live_df.shape)
```

```python
live_df = pd.read_csv("../data/weekly_trade_log_2026_week_5_with_auction_direction.csv")

live_df["Notes"] = (
    live_df["Notes"].fillna("").str.rstrip()
    + live_df["Unnamed: 9"].fillna("")
)

live_df = live_df.drop(columns=["Unnamed: 9"])
live_df = live_df[live_df["Date"] != "TOTAL"].copy()
live_df["Date"] = pd.to_datetime(live_df["Date"]).dt.strftime("%Y-%m-%d")
```

```python
combined_df = pd.concat(
    [historical_df, live_df],
    ignore_index=True
)

combined_df = combined_df.sort_values("Date").reset_index(drop=True)
```

```python
print("Shape:", combined_df.shape)
print("Date range:", combined_df["Date"].min(), "→", combined_df["Date"].max())
print("Duplicate dates:", combined_df["Date"].duplicated().sum())
```

```python
combined_df.to_csv(
    "../data/nq_first_to_100_predictions_full.csv",
    index=False
)
```
