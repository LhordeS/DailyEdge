```python
import pandas as pd

import matplotlib.pyplot as plt

df = pd.read_csv("../output/daily_stats.csv")
```

```python
df.groupby("Direction")[["Extension", "Counter"]].agg(
    Days=("Extension", "count"),
    AverageExtension=("Extension", "mean"),
    AverageCounter=("Counter", "mean"),
)
```

```python
df.groupby("Direction")[["Extension", "Counter"]].agg(
    Days=("Extension", "count"),
    AverageExtension=("Extension", "median"),
    AverageCounter=("Counter", "median"),
)
```

```python
thresholds = [35, 50, 70, 100, 150]

rows = []

for t in thresholds:
    bull = df[(df["Direction"] == "Bull") & (df["Counter"] <= t)]
    bear = df[(df["Direction"] == "Bear") & (df["Counter"] <= t)]

    bull_avg_ext = bull["Extension"].mean()
    bear_avg_ext = bear["Extension"].mean()

    rows.append({
        "Counter ≤": t,
        "Bull %": (len(bull) / len(df[df["Direction"] == "Bull"])) * 100,
        "Bull Avg Ext": bull_avg_ext,
        "Bull R:R": bull_avg_ext / t,
        "Bear %": (len(bear) / len(df[df["Direction"] == "Bear"])) * 100,
        "Bear Avg Ext": bear_avg_ext,
        "Bear R:R": bear_avg_ext / t,
    })

table = pd.DataFrame(rows).round(2)

table
```
